# Additive HSGP Residual Tuning

This notebook tests additive Hilbert-space Gaussian process residual models. Instead of one joint HSGP surface, it adds separate one-dimensional HSGP terms for boundary distance, evaluation progress, trajectory, and slope-instability coordinates. This follows the additive-GP idea before moving to HSGP residuals on top of Huber/GAM means or hierarchical GP models.

## 2. Slope-Core And Additive Coordinates

```python
"Define slope-core feature construction and deterministic sampling helpers."


def safe_numeric(series):
    return pd.to_numeric(series, errors="coerce").astype(float)


def deterministic_sample_positions(mask, max_rows):
    positions = np.flatnonzero(mask.to_numpy())
    if len(positions) <= max_rows:
        return positions
    return positions[np.linspace(0, len(positions) - 1, max_rows).astype(int)]


def deterministic_row_cap(frame, max_rows):
    if len(frame) <= max_rows:
        return frame
    positions = np.linspace(0, len(frame) - 1, max_rows).astype(int)
    return frame.iloc[positions].copy()


def window_slope(md, values, end_pos, window):
    start_pos = max(0, end_pos - window + 1)
    x0 = float(md.iloc[start_pos])
    x1 = float(md.iloc[end_pos])
    y0 = float(values.iloc[start_pos])
    y1 = float(values.iloc[end_pos])
    if not np.isfinite([x0, x1, y0, y1]).all() or abs(x1 - x0) < 1e-9:
        return np.nan
    return (y1 - y0) / (x1 - x0)


def build_slope_core_features_for_well(well, directory, max_rows):
    horizontal = load_horizontal_well(directory / f"{well}__horizontal_well.csv")
    target_mask = horizontal["TVT_input"].isna() & horizontal["TVT"].notna()
    known_mask = horizontal["TVT_input"].notna() & horizontal["TVT"].notna()
    if int(target_mask.sum()) == 0 or int(known_mask.sum()) == 0:
        return pd.DataFrame()

    tvt_input = safe_numeric(horizontal["TVT_input"])
    md = safe_numeric(horizontal["MD"])
    x = safe_numeric(horizontal["X"])
    y = safe_numeric(horizontal["Y"])
    z = safe_numeric(horizontal["Z"])

    known_positions = np.flatnonzero(known_mask.to_numpy())
    last_known_pos = int(known_positions[-1])
    last_known_tvt = float(tvt_input.iloc[last_known_pos])
    last_known_md = float(md.iloc[last_known_pos])
    last_known_z = float(z.iloc[last_known_pos])

    slope_windows = [10, 25, 50, 100]
    tvt_slopes = {f"tvt_slope_last_{window}": window_slope(md, tvt_input, last_known_pos, window) for window in slope_windows}
    z_slopes = {f"z_slope_last_{window}": window_slope(md, z, last_known_pos, window) for window in slope_windows}
    horizontal_step = np.sqrt(x.diff().fillna(0.0) ** 2 + y.diff().fillna(0.0) ** 2)
    horizontal_dist = horizontal_step.cumsum()

    target_positions = deterministic_sample_positions(target_mask, max_rows)
    all_target_positions = np.flatnonzero(target_mask.to_numpy())
    eval_rank = {int(pos): rank for rank, pos in enumerate(all_target_positions)}
    eval_denominator = max(len(all_target_positions) - 1, 1)

    rows = []
    for pos in target_positions:
        distance_md = float(md.iloc[pos] - last_known_md)
        row = {
            "wellname": well,
            "row_idx": int(horizontal.index[pos]),
            "md_norm": float((md.iloc[pos] - md.min()) / (md.max() - md.min() + 1e-9)),
            "distance_from_boundary_md": distance_md,
            "eval_progress": float(eval_rank[int(pos)] / eval_denominator),
            "last_known_tvt": last_known_tvt,
            "dZ_from_boundary": float(z.iloc[pos] - last_known_z),
            "horizontal_dist_from_boundary": float(horizontal_dist.iloc[pos] - horizontal_dist.iloc[last_known_pos]),
            "TVT": float(horizontal["TVT"].iloc[pos]),
            "endpoint_hold_pred": last_known_tvt,
        }
        row["delta_tvt_from_boundary"] = row["TVT"] - last_known_tvt
        for name, value in tvt_slopes.items():
            row[name] = float(value) if np.isfinite(value) else np.nan
            row[f"{name}_projected_delta"] = row[name] * distance_md if np.isfinite(row[name]) else np.nan
        for name, value in z_slopes.items():
            row[name] = float(value) if np.isfinite(value) else np.nan
        rows.append(row)
    return pd.DataFrame(rows)


slope_features = ["tvt_slope_last_10", "tvt_slope_last_25", "tvt_slope_last_50", "tvt_slope_last_100", "z_slope_last_25", "z_slope_last_50", "z_slope_last_100"]
slope_projection_features = ["tvt_slope_last_10_projected_delta", "tvt_slope_last_25_projected_delta", "tvt_slope_last_50_projected_delta", "tvt_slope_last_100_projected_delta"]
feature_columns = ["distance_from_boundary_md", "eval_progress", "md_norm", "dZ_from_boundary", "horizontal_dist_from_boundary"] + slope_features + slope_projection_features + ["last_known_tvt"]
coordinate_columns = ["distance_log1p", "eval_progress", "horizontal_log1p", "abs_projected_delta_10", "abs_dZ_from_boundary"]

print({"n_mean_features": len(feature_columns), "coordinate_columns": coordinate_columns})
```

```python
"Build sampled target-zone frame and additive HSGP coordinates."

start = time.time()
train_wells = discover_wells(TRAIN_DIR)
frames = [build_slope_core_features_for_well(well, TRAIN_DIR, MAX_TARGET_ROWS_PER_WELL) for well in train_wells]
feature_frame = pd.concat([frame for frame in frames if not frame.empty], ignore_index=True)
build_seconds = time.time() - start

feature_frame["distance_log1p"] = np.log1p(feature_frame["distance_from_boundary_md"].clip(lower=0.0))
feature_frame["horizontal_log1p"] = np.log1p(feature_frame["horizontal_dist_from_boundary"].clip(lower=0.0))
feature_frame["abs_projected_delta_10"] = feature_frame["tvt_slope_last_10_projected_delta"].abs()
feature_frame["abs_dZ_from_boundary"] = feature_frame["dZ_from_boundary"].abs()

print({"wells": len(train_wells), "sample_rows": len(feature_frame), "build_seconds": round(build_seconds, 2)})
feature_frame.head()
```

## 3. Additive HSGP Model Definitions

```python
"Define additive HSGP residual models and fold preparation helpers."


def linear_mu(X):
    n_features = X.shape[1]
    intercept = numpyro.sample("intercept", dist.Normal(0.0, 1.5))
    linear_beta = numpyro.sample("linear_beta", dist.Normal(0.0, 2.0).expand([n_features]).to_event(1))
    return intercept + jnp.dot(X, linear_beta)


def additive_hsgp_residual(X_gp, terms):
    residual_terms = []
    alpha_values = []
    length_values = []
    for idx, term in enumerate(terms):
        x_term = X_gp[:, term["column_index"]]
        alpha = numpyro.sample(f"hsgp_alpha_{idx}", dist.HalfNormal(term["alpha_scale"]))
        length = numpyro.sample(f"hsgp_length_{idx}", dist.LogNormal(term["length_log_mean"], term["length_log_sd"]))
        with handlers.scope(prefix=f"hsgp_{idx}"):
            if term["kernel"] == "matern32":
                residual = hsgp_matern(x=x_term, nu=1.5, alpha=alpha, length=length, ell=term["ell"], m=term["m"], non_centered=True)
            elif term["kernel"] == "matern52":
                residual = hsgp_matern(x=x_term, nu=2.5, alpha=alpha, length=length, ell=term["ell"], m=term["m"], non_centered=True)
            else:
                residual = hsgp_squared_exponential(x=x_term, alpha=alpha, length=length, ell=term["ell"], m=term["m"], non_centered=True)
        residual_terms.append(residual)
        alpha_values.append(alpha)
        length_values.append(length)
    if len(residual_terms) == 0:
        return jnp.zeros(X_gp.shape[0]), jnp.array([]), jnp.array([])
    return jnp.sum(jnp.stack(residual_terms), axis=0), jnp.stack(alpha_values), jnp.stack(length_values)


def student_t_linear_reference(X, X_gp, y=None, terms=None):
    mu = linear_mu(X)
    sigma = numpyro.sample("sigma", dist.HalfNormal(1.5))
    numpyro.deterministic("mu", mu)
    numpyro.deterministic("hsgp_residual", jnp.zeros(X.shape[0]))
    numpyro.sample("obs", dist.StudentT(3.0, mu, sigma), obs=y)


def student_t_additive_hsgp(X, X_gp, y=None, terms=None):
    linear_part = linear_mu(X)
    residual_gp, alpha_values, length_values = additive_hsgp_residual(X_gp, terms)
    mu = linear_part + residual_gp
    sigma = numpyro.sample("sigma", dist.HalfNormal(1.5))
    numpyro.deterministic("mu", mu)
    numpyro.deterministic("hsgp_residual", residual_gp)
    numpyro.deterministic("hsgp_alpha_values", alpha_values)
    numpyro.deterministic("hsgp_length_values", length_values)
    numpyro.sample("obs", dist.StudentT(3.0, mu, sigma), obs=y)


def term(column, m=8, ell=3.0, kernel="se", alpha_scale=0.55, length_log_mean=0.0, length_log_sd=0.45):
    return {"column": column, "m": m, "ell": ell, "kernel": kernel, "alpha_scale": alpha_scale, "length_log_mean": length_log_mean, "length_log_sd": length_log_sd}


candidate_specs = [
    {"candidate": "linear_t3_autodelta_reference", "model": student_t_linear_reference, "guide": "autodelta", "terms": []},
    {"candidate": "additive_dp_m8_se_autodelta", "model": student_t_additive_hsgp, "guide": "autodelta", "terms": [term("distance_log1p", 8), term("eval_progress", 8)]},
    {"candidate": "additive_dph_m8_se_autodelta", "model": student_t_additive_hsgp, "guide": "autodelta", "terms": [term("distance_log1p", 8), term("eval_progress", 8), term("horizontal_log1p", 8)]},
    {"candidate": "additive_dps_m8_se_autodelta", "model": student_t_additive_hsgp, "guide": "autodelta", "terms": [term("distance_log1p", 8), term("eval_progress", 8), term("abs_projected_delta_10", 8)]},
    {"candidate": "additive_dp_m10_matern32_autodelta", "model": student_t_additive_hsgp, "guide": "autodelta", "terms": [term("distance_log1p", 10, kernel="matern32"), term("eval_progress", 10, kernel="matern32")]},
]


def bind_terms(raw_terms):
    return [{**item, "column_index": coordinate_columns.index(item["column"])} for item in raw_terms]


def make_guide(spec):
    if spec["guide"] == "autonormal":
        return AutoNormal(spec["model"])
    return AutoDelta(spec["model"])


def fit_preprocessor(train_frame, columns):
    medians = train_frame[columns].median(numeric_only=True)
    filled = train_frame[columns].fillna(medians)
    means = filled.mean()
    stds = filled.std().replace(0.0, 1.0).fillna(1.0)
    return medians, means, stds


def apply_preprocessor(frame, columns, medians, means, stds):
    filled = frame[columns].fillna(medians)
    return ((filled - means) / stds).to_numpy(float)


def prepare_fold_arrays(train_frame, valid_frame):
    x_medians, x_means, x_stds = fit_preprocessor(train_frame, feature_columns)
    gp_medians, gp_means, gp_stds = fit_preprocessor(train_frame, coordinate_columns)
    X_train = apply_preprocessor(train_frame, feature_columns, x_medians, x_means, x_stds)
    X_valid = apply_preprocessor(valid_frame, feature_columns, x_medians, x_means, x_stds)
    Xgp_train = apply_preprocessor(train_frame, coordinate_columns, gp_medians, gp_means, gp_stds)
    Xgp_valid = apply_preprocessor(valid_frame, coordinate_columns, gp_medians, gp_means, gp_stds)
    y_train_raw = train_frame["delta_tvt_from_boundary"].to_numpy(float)
    y_mean = float(np.mean(y_train_raw))
    y_std = float(np.std(y_train_raw) + 1e-9)
    y_train = (y_train_raw - y_mean) / y_std
    return {"X_train": jnp.asarray(X_train), "X_valid": jnp.asarray(X_valid), "Xgp_train": jnp.asarray(Xgp_train), "Xgp_valid": jnp.asarray(Xgp_valid), "y_train": jnp.asarray(y_train), "y_mean": y_mean, "y_std": y_std}


def metric_row(y_true, pred_tvt):
    return {"rmse": float(np.sqrt(mean_squared_error(y_true, pred_tvt))), "mae": float(mean_absolute_error(y_true, pred_tvt)), "r2": float(r2_score(y_true, pred_tvt))}


def interval_metrics(y_true, draws):
    q10 = np.quantile(draws, 0.10, axis=0)
    q90 = np.quantile(draws, 0.90, axis=0)
    q025 = np.quantile(draws, 0.025, axis=0)
    q975 = np.quantile(draws, 0.975, axis=0)
    return {"coverage_80": float(((y_true >= q10) & (y_true <= q90)).mean()), "coverage_95": float(((y_true >= q025) & (y_true <= q975)).mean()), "width_80": float(np.mean(q90 - q10)), "width_95": float(np.mean(q975 - q025))}


pd.DataFrame([{"candidate": spec["candidate"], "guide": spec["guide"], "terms": [item["column"] for item in spec["terms"]], "basis_total": sum(item["m"] for item in spec["terms"]), "kernels": [item["kernel"] for item in spec["terms"]]} for spec in candidate_specs])
```

## 4. Grouped Additive HSGP CV

```python
"Run grouped well-fold validation for additive HSGP candidates."

fit_start = time.time()
fold_rows = []
residual_rows = []
posterior_rows = []
wells = sorted(feature_frame["wellname"].unique())
folds = group_well_folds(wells, n_splits=N_SPLITS)

for fold_idx, (train_fold_wells, valid_fold_wells) in enumerate(folds, start=1):
    train_frame = feature_frame.loc[feature_frame["wellname"].isin(train_fold_wells)].copy()
    valid_frame = feature_frame.loc[feature_frame["wellname"].isin(valid_fold_wells)].copy()
    train_frame = deterministic_row_cap(train_frame, MAX_TRAIN_ROWS_PER_FOLD)
    y_true_tvt = valid_frame["TVT"].to_numpy(float)
    endpoint_pred = valid_frame["endpoint_hold_pred"].to_numpy(float)
    arrays = prepare_fold_arrays(train_frame, valid_frame)
    fold_rows.append({"fold": fold_idx, "candidate": "endpoint_hold", "guide": "baseline", "term_columns": "baseline", "n_terms": 0, "basis_total": 0, "n_train_rows": 0, "n_valid_rows": len(valid_frame), "final_loss": np.nan, "coverage_80": np.nan, "coverage_95": np.nan, "width_80": np.nan, "width_95": np.nan, "hsgp_residual_sd": np.nan, **metric_row(y_true_tvt, endpoint_pred)})

    for spec_idx, spec in enumerate(candidate_specs):
        terms = bind_terms(spec["terms"])
        rng_key = jax.random.PRNGKey(SEED + fold_idx * 100 + spec_idx)
        guide = make_guide(spec)
        svi = SVI(spec["model"], guide, ClippedAdam(LEARNING_RATE, clip_norm=CLIP_NORM), Trace_ELBO())
        svi_result = svi.run(rng_key, SVI_STEPS, arrays["X_train"], arrays["Xgp_train"], arrays["y_train"], terms, progress_bar=False)
        params = svi_result.params
        final_loss = float(svi_result.losses[-1])
        pred_key = jax.random.PRNGKey(SEED + fold_idx * 1000 + spec_idx)
        predictive = Predictive(spec["model"], guide=guide, params=params, num_samples=POSTERIOR_SAMPLES, return_sites=("mu", "obs", "hsgp_residual", "hsgp_alpha_values", "hsgp_length_values"))
        pred = predictive(pred_key, arrays["X_valid"], arrays["Xgp_valid"], None, terms)
        mu_draws_delta = np.asarray(pred["mu"]) * arrays["y_std"] + arrays["y_mean"]
        obs_draws_delta = np.asarray(pred["obs"]) * arrays["y_std"] + arrays["y_mean"]
        hsgp_draws_delta = np.asarray(pred["hsgp_residual"]) * arrays["y_std"]
        last_known = valid_frame["last_known_tvt"].to_numpy(float)[None, :]
        tvt_mu_draws = mu_draws_delta + last_known
        tvt_obs_draws = obs_draws_delta + last_known
        pred_tvt = tvt_mu_draws.mean(axis=0)
        if not np.isfinite(pred_tvt).all() or not np.isfinite(tvt_obs_draws).all():
            print({"nonfinite_candidate": spec["candidate"], "fold": fold_idx, "guide": spec["guide"], "terms": terms, "final_loss": final_loss, "pred_finite": bool(np.isfinite(pred_tvt).all()), "obs_finite": bool(np.isfinite(tvt_obs_draws).all())})
            raise ValueError("Non-finite additive HSGP predictions")
        metrics = metric_row(y_true_tvt, pred_tvt)
        intervals = interval_metrics(y_true_tvt, tvt_obs_draws)
        term_columns = "+".join(item["column"] for item in terms) if terms else "none"
        basis_total = int(sum(item["m"] for item in terms))
        hsgp_residual_sd = float(hsgp_draws_delta.mean(axis=0).std())
        alpha_values = np.asarray(pred.get("hsgp_alpha_values", np.empty((POSTERIOR_SAMPLES, 0))))
        length_values = np.asarray(pred.get("hsgp_length_values", np.empty((POSTERIOR_SAMPLES, 0))))
        fold_rows.append({"fold": fold_idx, "candidate": spec["candidate"], "guide": spec["guide"], "term_columns": term_columns, "n_terms": len(terms), "basis_total": basis_total, "n_train_rows": len(train_frame), "n_valid_rows": len(valid_frame), "final_loss": final_loss, "hsgp_residual_sd": hsgp_residual_sd, **metrics, **intervals})
        posterior_rows.append({"fold": fold_idx, "candidate": spec["candidate"], "final_loss": final_loss, "term_columns": term_columns, "basis_total": basis_total, "hsgp_residual_sd": hsgp_residual_sd, "alpha_mean": float(np.nanmean(alpha_values)) if alpha_values.size else np.nan, "length_mean": float(np.nanmean(length_values)) if length_values.size else np.nan})
        sample_idx = np.linspace(0, len(valid_frame) - 1, min(len(valid_frame), RESIDUAL_SAMPLE_PER_FOLD)).astype(int)
        residual_rows.append(pd.DataFrame({"fold": fold_idx, "candidate": spec["candidate"], "wellname": valid_frame.iloc[sample_idx]["wellname"].to_numpy(), "row_idx": valid_frame.iloc[sample_idx]["row_idx"].to_numpy(), "distance_from_boundary_md": valid_frame.iloc[sample_idx]["distance_from_boundary_md"].to_numpy(float), "eval_progress": valid_frame.iloc[sample_idx]["eval_progress"].to_numpy(float), "horizontal_dist_from_boundary": valid_frame.iloc[sample_idx]["horizontal_dist_from_boundary"].to_numpy(float), "abs_projected_delta_10": valid_frame.iloc[sample_idx]["abs_projected_delta_10"].to_numpy(float), "y_true": y_true_tvt[sample_idx], "y_pred": pred_tvt[sample_idx], "residual": y_true_tvt[sample_idx] - pred_tvt[sample_idx], "hsgp_residual_mean": hsgp_draws_delta[:, sample_idx].mean(axis=0), "posterior_sd": tvt_obs_draws[:, sample_idx].std(axis=0)}))

fit_seconds = time.time() - fit_start
fold_metrics = pd.DataFrame(fold_rows)
summary = (fold_metrics.groupby(["candidate", "guide", "term_columns", "n_terms", "basis_total"], as_index=False).agg(rmse_mean=("rmse", "mean"), rmse_std=("rmse", "std"), mae_mean=("mae", "mean"), r2_mean=("r2", "mean"), coverage_80_mean=("coverage_80", "mean"), coverage_95_mean=("coverage_95", "mean"), width_80_mean=("width_80", "mean"), width_95_mean=("width_95", "mean"), hsgp_residual_sd_mean=("hsgp_residual_sd", "mean"), n_valid_rows=("n_valid_rows", "sum"), n_train_rows_mean=("n_train_rows", "mean"), final_loss_mean=("final_loss", "mean")).sort_values("rmse_mean"))
residual_sample = pd.concat(residual_rows, ignore_index=True)
posterior_diagnostics = pd.DataFrame(posterior_rows)

print({"fit_seconds": round(fit_seconds, 2), "fold_rows": len(fold_metrics), "residual_rows": len(residual_sample)})
summary
```
